```python
import pandas as pd
from pgmpy.models import BayesianNetwork
from pgmpy.estimators import MaximumLikelihoodEstimator
from pgmpy.inference import VariableElimination

df = pd.read_csv('wbes_sintetico.csv').dropna()

promedio_ventas = df['sales_annual_usd'].mean()
df['successful'] = (df['sales_annual_usd'] > promedio_ventas).astype(int)

df['training'] = df['training'].astype(int)
df['access_finance'] = df['access_finance'].astype(int)
df['exporter'] = df['exporter'].astype(int)

variables_red = ['successful', 'training', 'access_finance', 'competition_level', 'exporter']
df_bayes = df[variables_red]

modelo_bayes = BayesianNetwork([
    ('access_finance', 'successful'),
    ('training', 'successful'),
    ('competition_level', 'successful'),
    ('exporter', 'successful')
])

modelo_bayes.fit(df_bayes, estimator=MaximumLikelihoodEstimator)

inferencia = VariableElimination(modelo_bayes)
probabilidad = inferencia.query(variables=['successful'], evidence={'training': 1, 'access_finance': 1})

print(probabilidad)
```
