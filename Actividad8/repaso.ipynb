```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# 1. Crear el DataFrame con los datos proporcionados
data = {
    'Dia': list(range(1, 26)),
    'Ingreso': [
        1452, 1361, 1426, 1470, 1456, 1430, 1354, 1442, 1394, 1459, 1399, 1458, 1537,
        1425, 1445, 1439, 1348, 1450, 1431, 1446, 1485, 1405, 1461, 1490, 1426
    ],
    'Habitaciones': [
        23, 47, 21, 39, 37, 29, 23, 44, 45, 16, 30, 42, 54,
        27, 34, 15, 19, 38, 44, 47, 43, 38, 51, 61, 39
    ]
}
```

```python
df = pd.DataFrame(data)

# 2. Gráfica de dispersión (Scatter plot)
sns.lmplot(x='Habitaciones', y='Ingreso', data=df, height=5, aspect=1.5)
plt.title('Correlación Lineal: Habitaciones vs. Ingreso')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# 3. Cálculo del coeficiente de correlación de Pearson
correlacion = df['Habitaciones'].corr(df['Ingreso'])
print(f"Coeficiente de correlación de Pearson: {correlacion:.4f}")
```

# 2. Interpretación de ResultadosAnálisis Visual: 
 La gráfica muestra una línea de tendencia ascendente, lo que indica que existe una relación positiva: a mayor número de habitaciones ocupadas, el ingreso del restaurante tiende a aumentar.Coeficiente (r = 0.42): El cálculo arroja un valor de 0.42. Esto se interpreta como una correlación positiva moderada. Aunque las variables están relacionadas, la dispersión de los puntos sugiere que la ocupación del hotel no es el único factor que define el éxito del restaurante.
