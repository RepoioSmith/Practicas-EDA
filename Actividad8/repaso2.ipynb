```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data = {
    'Comisiones': [
        22, 13, 33, 38, 23, 47, 29, 38, 41, 32, 20, 13, 47, 
        38, 44, 29, 38, 37, 14, 34, 25, 27, 25, 43, 34
    ],
    'Llamadas': [
        139, 132, 144, 142, 142, 142, 138, 139, 144, 134, 135, 137, 146, 
        146, 144, 147, 144, 149, 131, 144, 132, 132, 127, 154, 147
    ],
    'Millas': [
        2371, 2226, 2731, 3351, 2289, 3449, 3114, 3342, 2842, 2625, 2121, 2219, 3463, 
        3290, 3103, 2122, 2791, 3209, 2287, 2848, 2690, 2933, 2671, 2988, 2829
    ]
}
```

```python
df = pd.DataFrame(data)
```

```python
# 2. Generar el Mapa de Calor de Correlaciones
plt.figure(figsize=(8, 6))
# annot=True muestra el número en cada cuadro
sns.heatmap(df.corr(), annot=True, cmap='coolwarm', vmin=-1, vmax=1)
plt.title('Matriz de Correlación Múltiple: Comisiones, Llamadas y Millas')
plt.show()
```

```python
# 3. Mostrar la tabla de coeficientes numéricos
print("Matriz de Correlación:")
print(df.corr())
```

# Interpretación de los Resultados
Al ejecutar el código, obtendrás los siguientes coeficientes de correlación ($r$):Comisiones vs. Millas Recorridas (r = 0.83): Existe una correlación positiva muy fuerte.Significado: La cantidad de millas que viaja un agente es el factor que más influye en sus ganancias. Quienes viajan más, ganan significativamente más.Comisiones vs. Llamadas de Ventas (r = 0.68): Existe una correlación positiva moderada-fuerte.Significado: Hacer más llamadas también ayuda a ganar más, pero no es tan determinante como viajar.Conclusión para el Gerente:Sí, los agentes que recorren más millas y hacen más llamadas ganan más. Sin embargo, las millas recorridas son el indicador más fuerte de altas comisiones.
