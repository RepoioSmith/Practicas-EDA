```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

#Importe de Datos
df = pd.read_csv("./SalsberryRealty.csv")
df.head()
```

## Determinar las variables independientes

```python
X = df[['Temp', 'Insul', 'Age']]
X
```

## Determinar la variables dependiente

```python
y =df[['Cost']]
y
```

## Ajustar el modelo para encontrar el modelo de regresion multiple

```python
model = LinearRegression()
model.fit(X,y)
y_pred = model.predict(X)

mse = mean_squared_error(y,y_pred)
r2 = r2_score(y,y_pred)
```

## Obtener el valor de interseccion

```python
model.intercept_
```

```python
print('Coefcientes: ', model.coef_)
print('Error cuadratico de la media', mse)
print('Variaza: ', r2)
```

## 4. Estimar para 30, aislamiento 5", antiguedad 10 años

```python
Y = model.predict([[30, 5, 10]])
print("El costo de calefaccion es: ", Y)
```
